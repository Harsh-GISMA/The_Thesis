# Import required libraries

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from google.colab import drive
```

# 1. Mount Google Drive

```python
drive.mount('/content/drive')
```

# 2. Define file paths

```python
master_file_path = '/content/drive/MyDrive/Thesis/Dataset/results_master_stacked.csv'
output_file_path = '/content/drive/MyDrive/Thesis/Dataset/visual_au_processed.csv'
```

# 3. Load the master dataset

```python
df_master = pd.read_csv(master_file_path)
```

# 4. Define the target Image/Visual columns (14 Facial AUs)

```python
au_columns = [
    'AU01_r_mean', 'AU01_r_std',
    'AU02_r_mean', 'AU02_r_std',
    'AU04_r_mean', 'AU04_r_std',
    'AU06_r_mean', 'AU06_r_std',
    'AU12_r_mean', 'AU12_r_std',
    'AU14_r_mean', 'AU14_r_std',
    'AU15_r_mean', 'AU15_r_std'
]
df_visual = df_master[['participant_id'] + au_columns].copy()
```

# 5. Clean & Impute (Median strategy for robustness against outliers)

```python
imputer = SimpleImputer(strategy='median')
df_visual[au_columns] = imputer.fit_transform(df_visual[au_columns])
```

# 6. Z-score Normalization

```python
scaler = StandardScaler()
df_visual[au_columns] = scaler.fit_transform(df_visual[au_columns])
```

# 7. Save the processed visual modality matrix

```python
df_visual.to_csv(output_file_path, index=False)
print(f"Visual AU processing complete. Shape: {df_visual.shape}")
print("Sample of processed visual data:")
print(df_visual.head())
```

## Pipeline Summary

This notebook performs the following steps to process the visual modality data:

1.  **Mount Google Drive**: Accesses files stored in Google Drive.
2.  **Define File Paths**: Specifies the input master dataset and the output path for the processed visual data.
3.  **Load Master Dataset**: Reads the `results_master_stacked.csv` file into a pandas DataFrame.
4.  **Define Target Columns**: Selects specific facial Action Units (AUs) and `participant_id` to create `df_visual`.
5.  **Clean & Impute**: Uses `SimpleImputer` with a 'median' strategy to handle missing values in the AU columns.
6.  **Z-score Normalization**: Applies `StandardScaler` to perform Z-score normalization on the AU columns.
7.  **Save Processed Data**: Saves the cleaned and normalized visual AU data to `visual_au_processed.csv`.

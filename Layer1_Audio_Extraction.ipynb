# Importing all the Libreraies


```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

# 1. Load Data

```python
df = pd.read_csv("/content/drive/MyDrive/Thesis/Dataset/results_master_stacked.csv")
```

# 2. Hardcode your manually verified columns

```python
id_col = 'participant_id'
covarep_cols = [
    'F0_mean',
    'voiced_ratio',
    'F0_std',
    'NAQ_mean',
    'NAQ_std',
    'H1H2_mean',
    'H1H2_std',
    'peakSlope_mean',
    'peakSlope_std',
    'Rd_mean',
    'Rd_std'
]
X_acoustic = df[covarep_cols].copy()
```

# 3. Clean and Scale (Imputation handles missing DAIC-WOZ clinical frames)

```python
X_acoustic.replace([np.inf, -np.inf], np.nan, inplace=True)
X_imputed = SimpleImputer(strategy="median").fit_transform(X_acoustic)
X_scaled = StandardScaler().fit_transform(X_imputed)
```

# 4. Reconstruct and Save for Multimodal Fusion

```python
df_processed = pd.DataFrame(X_scaled, columns=covarep_cols)
df_processed.insert(0, id_col, df[id_col].values)

df_processed.to_csv("/content/drive/MyDrive/Thesis/Dataset/voice_covarep_processed.csv", index=False)
print(f"Successfully processed {len(df_processed)} participants across {len(covarep_cols)} acoustic features.")
```

## Pipeline Summary

This notebook processes acoustic features (COVAREP) from a raw dataset. It performs the following steps:

1.  **Loads the raw data** from `results_master_stacked.csv`.
2.  **Selects specific COVAREP columns** for processing.
3.  **Cleans and Scales the data** by:
    *   Replacing infinite values with NaN.
    *   Imputing missing values using a median strategy.
    *   Scaling the features using `StandardScaler`.
4.  **Reconstructs the processed data** into a new DataFrame, including the participant ID.
5.  **Saves the processed data** to `/content/drive/MyDrive/Thesis/Dataset/voice_covarep_processed.csv` for multimodal fusion.
